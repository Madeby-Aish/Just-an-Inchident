# Import the Packages

```python
import gradio
from groq import Groq #Groq class from groq
```

```python
client = Groq(api_key="API key herex") #object called client with api key as string
```

#Define a function to give content and role

```python
def initialize_messages():
    return [{"role": "system",
             "content": """You are a formula 1 race strategist with a
             successful track record in race strategies. Your role is to
             assist people by providing guidance on race, strategies, and
             offering answers in a professional manner."""}]
```

#Assign it to a variable

```python
messages_prmt = initialize_messages()
```

#Define a function to connect with LLM

```python
def customLLMBot(user_input, history):
    global messages_prmt

    messages_prmt.append({"role": "user", "content": user_input})

    response = client.chat.completions.create(
        messages=messages_prmt,
        model="openai/gpt-oss-120b",
    )
    print(response)
    LLM_reply = response.choices[0].message.content
    messages_prmt.append({"role": "assistant", "content": LLM_reply})

    return LLM_reply
```

#Create an object of chat interface class in gradio

```python
iface = gradio.ChatInterface(customLLMBot,
                     chatbot=gradio.Chatbot(height=300),
                     textbox=gradio.Textbox(placeholder="Ask me a question related to Formula1 race strategy"),
                     title="Just an 'Inchident' ",
                     description="Chat bot for one stop assistance tailored for information reguarding strategies in Formula 1",
                     examples=["Hi","what is Formula 1", "simple instance of a race strategies reguarding pit stops"]
                     )
```

#Call launch function to execute

```python
iface.launch(share=True)
```
